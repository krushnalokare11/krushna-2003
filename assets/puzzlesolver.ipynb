```python
from collections import deque
import heapq

# ---------------- Utility Functions ----------------

def print_state(state):
    """Display the 3x3 puzzle neatly."""
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()

def manhattan_distance(state, goal):
    """Calculate Manhattan Distance heuristic."""
    distance = 0
    for i, val in enumerate(state):
        if val == 0:
            continue
        goal_index = goal.index(val)
        x1, y1 = divmod(i, 3)
        x2, y2 = divmod(goal_index, 3)
        distance += abs(x1 - x2) + abs(y1 - y2)
    return distance

# ---------------- Move Generator ----------------

def get_neighbors(state):
    """Generate neighboring states by moving the blank (0)."""
    neighbors = []
    idx = state.index(0)
    x, y = divmod(idx, 3)
    moves = {'up': (-1, 0), 'down': (1, 0), 'left': (0, -1), 'right': (0, 1)}
    for move, (dx, dy) in moves.items():
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = state.copy()
            new_idx = nx * 3 + ny
            new_state[idx], new_state[new_idx] = new_state[new_idx], new_state[idx]
            neighbors.append((move, new_state))
    return neighbors

# ---------------- BFS Implementation ----------------

def bfs(start, goal):
    queue = deque([(start, [])])
    visited = set()
    step = 0

    while queue:
        state, path = queue.popleft()
        visited.add(tuple(state))
        step += 1
        print(f"Step {step}: {path[-1] if path else 'Start'}")
        print_state(state)
        print(f"Queue length: {len(queue)} | Visited: {len(visited)}\n")

        if state == goal:
            print(f"Goal reached in {len(path)} moves: {path}")
            return path

        for move, neighbor in get_neighbors(state):
            if tuple(neighbor) not in visited:
                queue.append((neighbor, path + [move]))

    print("Goal not found.")

# ---------------- A* Implementation ----------------

def astar(start, goal):
    open_list = []
    heapq.heappush(open_list, (manhattan_distance(start, goal), 0, start, []))
    visited = set()
    step = 0

    while open_list:
        f, g, state, path = heapq.heappop(open_list)
        if tuple(state) in visited:
            continue
        visited.add(tuple(state))
        step += 1

        print(f"Step {step}: {path[-1] if path else 'Start'} (f={f}, g={g})")
        print_state(state)
        print(f"Open list size: {len(open_list)} | Visited: {len(visited)}\n")

        if state == goal:
            print(f"Goal reached in {len(path)} moves: {path}")
            return path

        for move, neighbor in get_neighbors(state):
            if tuple(neighbor) not in visited:
                g_new = g + 1
                f_new = g_new + manhattan_distance(neighbor, goal)
                heapq.heappush(open_list, (f_new, g_new, neighbor, path + [move]))

    print("Goal not found.")

# ---------------- Custom Free-Move Function ----------------

def move_any_tile(state, tile, direction):
    """Allow any tile (1–8) to move freely in the given direction."""
    if tile not in state or tile == 0:
        print("Invalid tile number. Choose between 1–8.")
        return state

    index = state.index(tile)
    x, y = divmod(index, 3)

    moves = {"up": (-1, 0), "down": (1, 0), "left": (0, -1), "right": (0, 1)}

    if direction not in moves:
        print("Invalid direction. Use up/down/left/right.")
        return state

    dx, dy = moves[direction]
    nx, ny = x + dx, y + dy

    # Ensure move stays inside the 3x3 grid
    if not (0 <= nx < 3 and 0 <= ny < 3):
        print("Move out of bounds! Try another direction.")
        return state

    new_state = state.copy()
    new_index = nx * 3 + ny
    new_state[index], new_state[new_index] = new_state[new_index], new_state[index]
    return new_state

# ---------------- Interactive Mode ----------------

def interactive_mode():
    state = [1, 2, 5, 3, 4, 0, 6, 7, 8]
    goal = [1, 2, 3, 4, 5, 6, 7, 8, 0]

    print("Initial State:")
    print_state(state)

    while True:
        cmd = input("Enter 'move', 'auto bfs', 'auto astar', or 'quit': ").lower()

        if cmd == 'quit':
            break
        elif cmd == 'auto bfs':
            bfs(state, goal)
        elif cmd == 'auto astar':
            astar(state, goal)
        elif cmd == 'move':
            try:
                tile = int(input("Enter the tile number (1–8) you want to move: "))
                direction = input("Enter direction (up/down/left/right): ").lower()
                new_state = move_any_tile(state, tile, direction)
                print(f"\nMoved tile {tile} {direction}:")
                print_state(new_state)
                state = new_state
            except ValueError:
                print("Please enter a valid number.\n")
        else:
            print("Unknown command.\n")

if __name__ == "__main__":
    interactive_mode()
```
